```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller
from sklearn.metrics import mean_absolute_error, mean_squared_error
import os

def make_stationary(time_series):
    """Applies differencing iteratively until the series becomes stationary."""
    diff_count = 0
    while True:
        adf_test = adfuller(time_series.dropna())
        if adf_test[1] <= 0.05:
            break
        time_series = time_series.diff().dropna()
        diff_count += 1
    return time_series, diff_count

def calculate_metrics(actual, predicted):
    """Calculate MAE, RMSE, and MAPE."""
    epsilon = 1e-5  # To avoid division by zero
    actual = np.array(actual)
    predicted = np.array(predicted)
    
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mape = np.mean(np.abs((actual - predicted) / np.maximum(actual, epsilon))) * 100
    
    return mae, rmse, mape

def analyze_entity_rolling(entity, data, train_ratio=0.8, rolling_window=5):
    entity_data = data[data['Entity'] == entity].copy()
    entity_data['Year'] = pd.to_datetime(entity_data['Year'], format='%Y')
    entity_data.set_index('Year', inplace=True)
    time_series = entity_data['Coal Consumption - TWh']
    
    if len(time_series) < 20:
        print(f"Not enough data for {entity}. Please select another country.")
        return
    
    time_series_stationary, d = make_stationary(time_series)
    
    train_size = int(len(time_series) * train_ratio)
    
    arima_errors, sarima_errors = [], []
    
    for i in range(rolling_window, len(time_series) - train_size):
        train_data = time_series.iloc[:train_size + i]
        test_data = time_series.iloc[train_size + i: train_size + i + rolling_window]
        
        if len(test_data) == 0:
            break
            
        # Create directories if they don't exist
        os.makedirs(f"Training Data/{entity}", exist_ok=True)
        os.makedirs(f"Testing Data/{entity}", exist_ok=True)
        
        # Save training and testing data in respective folders WITH YEARS
        train_data.to_csv(f"Training Data/{entity}/train_data_{entity}_{i}.csv", index=True)  # Include years
        test_data.to_csv(f"Testing Data/{entity}/test_data_{entity}_{i}.csv", index=True)  # Include years
        
        p, q = 1, 1
        arima_model = SARIMAX(train_data, order=(p, d, q), seasonal_order=(0, 0, 0, 0))
        arima_results = arima_model.fit()
        arima_forecast = arima_results.get_forecast(steps=len(test_data)).predicted_mean
        
        P, Q, s = 1, 1, 12  # Assuming yearly seasonality
        sarima_model = SARIMAX(train_data, order=(p, d, q), seasonal_order=(P, d, Q, s))
        sarima_results = sarima_model.fit()
        sarima_forecast = sarima_results.get_forecast(steps=len(test_data)).predicted_mean
        
        arima_errors.append(calculate_metrics(test_data, arima_forecast))
        sarima_errors.append(calculate_metrics(test_data, sarima_forecast))
    
    arima_errors = np.array(arima_errors)
    sarima_errors = np.array(sarima_errors)
    # Calculate Mean Values
    arima_mae, arima_rmse, arima_mape = np.mean(arima_errors, axis=0)
    sarima_mae, sarima_rmse, sarima_mape = np.mean(sarima_errors, axis=0)
    
    print(f"ARIMA Metrics: MAE: {arima_mae}, RMSE: {arima_rmse}, MAPE: {arima_mape}%")
    print(f"SARIMA Metrics: MAE: {sarima_mae}, RMSE: {sarima_rmse}, MAPE: {sarima_mape}%")
    if arima_mape < sarima_mape:
        print("ARIMA performed better.")
    else:
        print("SARIMA performed better.")
    
    # Plot Accuracy Metrics
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.plot(arima_errors[:, 0], label="ARIMA MAE", marker='o')
    plt.plot(sarima_errors[:, 0], label="SARIMA MAE", marker='s')
    plt.xlabel("Rolling Iteration")
    plt.ylabel("MAE")
    plt.title("MAE Comparison")
    plt.legend()
    plt.subplot(1, 3, 2)
    plt.plot(arima_errors[:, 1], label="ARIMA RMSE", marker='o')
    plt.plot(sarima_errors[:, 1], label="SARIMA RMSE", marker='s')
    plt.xlabel("Rolling Iteration")
    plt.ylabel("RMSE")
    plt.title("RMSE Comparison")
    plt.legend()
    plt.subplot(1, 3, 3)
    plt.plot(arima_errors[:, 2], label="ARIMA MAPE", marker='o')
    plt.plot(sarima_errors[:, 2], label="SARIMA MAPE", marker='s')
    plt.xlabel("Rolling Iteration")
    plt.ylabel("MAPE (%)")
    plt.title("MAPE Comparison")
    plt.legend()
    plt.tight_layout()
    plt.show()
    # Print final accuracy metrics after all iterations
    print("\nFinal ARIMA Accuracy:")
    print(f"MAE: {arima_mae:.4f}, RMSE: {arima_rmse:.4f}, MAPE: {arima_mape:.2f}%")
    print("\nFinal SARIMA Accuracy:")
    print(f"MAE: {sarima_mae:.4f}, RMSE: {sarima_rmse:.4f}, MAPE: {sarima_mape:.2f}%")
    # Print final accuracy metrics after all iterations
    print("\nFinal ARIMA Accuracy:")
    print(f"MAE: {arima_mae:.4f}, RMSE: {arima_rmse:.4f}, MAPE: {arima_mape:.2f}%")
    print(f"Accuracy: {100 - arima_mape:.2f}%")
    print("\nFinal SARIMA Accuracy:")
    print(f"MAE: {sarima_mae:.4f}, RMSE: {sarima_rmse:.4f}, MAPE: {sarima_mape:.2f}%")
    print(f"Accuracy: {100 - sarima_mape:.2f}%")

def predict_future(entity, data, years=10):
    entity_data = data[data['Entity'] == entity].copy()
    entity_data['Year'] = pd.to_datetime(entity_data['Year'], format='%Y')
    entity_data.set_index('Year', inplace=True)
    time_series = entity_data['Coal Consumption - TWh']
    
    if len(time_series) < 20:
        print(f"Not enough data for {entity}. Please select another country.")
        return
    
    time_series_stationary, d = make_stationary(time_series)
    p, q = 1, 1
    P, Q, s = 1, 1, 12  # Assuming yearly seasonality
    
    # Train ARIMA model
    arima_model = SARIMAX(time_series, order=(p, d, q), seasonal_order=(0, 0, 0, 0))
    arima_results = arima_model.fit()
    
    # Train SARIMA model
    sarima_model = SARIMAX(time_series, order=(p, d, q), seasonal_order=(P, d, Q, s))
    sarima_results = sarima_model.fit()
    
    # Generate forecasts
    arima_forecast = arima_results.get_forecast(steps=years).predicted_mean
    sarima_forecast = sarima_results.get_forecast(steps=years).predicted_mean
    
    # Use recent historical data to compare both models
    test_size = min(5, len(time_series))  # Compare with the last 5 years or less if data is small
    actual_recent = time_series[-test_size:]
    arima_recent_forecast = arima_results.get_forecast(steps=test_size).predicted_mean
    sarima_recent_forecast = sarima_results.get_forecast(steps=test_size).predicted_mean
    
    # Compute MAPE for both models on recent data
    arima_mae, arima_rmse, arima_mape = calculate_metrics(actual_recent, arima_recent_forecast)
    sarima_mae, sarima_rmse, sarima_mape = calculate_metrics(actual_recent, sarima_recent_forecast)
    
    # Select the best model
    if arima_mape < sarima_mape:
        best_forecast = arima_forecast
        model_name = "ARIMA"
    else:
        best_forecast = sarima_forecast
        model_name = "SARIMA"
    
    # Extend the index for the next years
    future_dates = pd.date_range(start=time_series.index[-1] + pd.DateOffset(years=1), periods=years, freq='Y')
    
    # Create a DataFrame for the predictions
    predictions_df = pd.DataFrame({
        'Year': future_dates.year,
        'Predicted Coal Consumption (TWh)': best_forecast
    })
    
    # Save predictions to a CSV file in the "Predictions/{entity}" folder
    os.makedirs(f"Predictions/{entity}", exist_ok=True)  # Create the country-specific folder if it doesn't exist
    predictions_df.to_csv(f"Predictions/{entity}/{entity}_predictions.csv", index=False)
    
    # Plot results
    plt.figure(figsize=(12, 6))
    plt.plot(time_series.index, time_series, label="Actual Data", marker='o', linestyle='-')
    plt.plot(future_dates, best_forecast, label=f"Predicted ({model_name})", marker='s', linestyle='dashed', color='red')
    plt.xlabel("Year")
    plt.ylabel("Coal Consumption (TWh)")
    plt.title(f"Coal Consumption Forecast for {entity} (Next {years} Years)")
    plt.legend()
    plt.grid()
    plt.show()
    
    print(f"\nFuture Predictions ({model_name} Model):")
    for date, value in zip(future_dates, best_forecast):
        print(f"{date.year}: {value:.2f} TWh")
    
# Load dataset
data = pd.read_excel(r"C:\Users\KARTHIKEYA\Desktop\Mini Project\Datasets\COAL Consumption.xlsx", 
                     sheet_name="coal-consumption-by-country-ter")
# Define Country (eg: India)
country = "Africa"
#Train Models (based on rolling window data splitting technique)
analyze_entity_rolling(country, data)
# Predict for the next 10 years
predict_future(country, data)
```
